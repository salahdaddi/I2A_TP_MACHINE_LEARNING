*Ce notebook a pour objectif d'étudier et de visualiser différentes fonctions d'activation utilisées dans les neurones formels et les réseaux de neurones artificiels. À l'aide des bibliothèques NumPy et Matplotlib, un intervalle de valeurs est généré puis plusieurs fonctions d'activation peuvent être sélectionnées par l'utilisateur : Step, Linear, ReLU, Sigmoid et TanH. Pour chaque fonction choisie, le programme calcule la sortie correspondante et affiche sa représentation graphique. Cette expérimentation permet de comparer le comportement de ces fonctions et de mieux comprendre leur rôle dans les modèles d'apprentissage automatique.*

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x= np.arange(-10,10,0.1)
n = len(x)
```

```python
def fct ():
    function = input('Input the activation function. Example : Step, Linear, ReLU, Sigmoid, TanH :')
    if function == 'Step' :
        y = np.zeros(n)
        for i in range(n):
            if x[i]< 0 :
                y[i] = 0
            else :
                y[i] = 1
        plt.plot(x,y)
    elif function == 'Linear' :
        y = x.copy() 
        plt.plot(x,y)
    elif function == 'ReLU' :
        y = x.copy()
        for i in range(n):
            if x[i]< 0 :
                y[i] = 0
        plt.plot(x,y)
    elif function == 'Sigmoid' : 
        y = 1/(1+np.exp(-x))
        plt.plot(x,y)
    elif function == 'TanH' :
        y = (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))
        plt.plot(x,y)
    else : 
        print('Not reconized function')
        
    
                
```

```python
fct()
```
