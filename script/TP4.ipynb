```python
import numpy as np
```

```python

# Inputs
weather = int(input("Enter weather (0 or 1): "))
distance = int(input("Enter distance (0 or 1): "))
company = int(input("Enter company (0 or 1): "))

bias = -0.5
w1 = 0.6
w2 = 0.2
w3 = 0.2

# Calculate perceptron output
z = w1 * weather + w2 * distance + w3 * company + bias 

# Activation function
if z > 0:
    print("Go to the library")
else:
    print("Do not go to the library")
```

```python
# Inputs
weather = int(input("Enter weather (0 or 1): "))
distance = int(input("Enter distance (0 or 1): "))
company = int(input("Enter company (0 or 1): "))

bias = -0.5
w1 = 0.5
w2 = 0.2
w3 = 0.2

# Calculate perceptron output
z = w1 * weather + w2 * distance + w3 * company + bias 
sigmoid = 1/(1+ np.exp(-z))

# Activation function
if sigmoid > 0.5:
    print("Go to the library")
else:
    print("Do not go to the library")
```
